## 3  Imports


```python
from circuitkit import Pipeline
from circuitkit.applications.editing.knowledge_editing import CircuitKnowledgeEditor
```

## 4  Config

Edit this cell. Everything below reads these names.


```python
MODEL     = "Qwen/Qwen2.5-0.5B-Instruct"
TASK      = "ioi"
PRECISION = "bfloat16"
DEVICE    = "cuda"
OUT_DIR   = "./ck_out"
N_EXAMPLES = 32
BATCH_SIZE = 1
SPARSITY   = 0.3
SCOPE      = "both"   # heads | mlp | both
LEVEL      = "node"   # node | neuron
SEED       = 0
ALGORITHM = "eap-ig"
PROMPT  = "The capital of France is"
SUBJECT = "France"
TARGET  = "Lyon"
VERIFY = True
N_PREFIXES = 2
COV_N_SAMPLES = 8
```

## 5  Build


```python
pipe = Pipeline(
    MODEL,
    task=TASK,
    precision=PRECISION,
    device=DEVICE,
    output_dir=OUT_DIR,
)
pipe.discover(algorithm=ALGORITHM, n_examples=N_EXAMPLES, batch_size=BATCH_SIZE)
circuit = pipe._circuit
editor = CircuitKnowledgeEditor(pipe._ensure_model())
```

## 6  Run


```python
result = editor.edit_via_circuit(
    prompt=PROMPT,
    subject=SUBJECT,
    target=TARGET,
    circuit=circuit,
    method="rome",
    verify=VERIFY,
    n_prefixes=N_PREFIXES,
    cov_n_samples=COV_N_SAMPLES,
)
print(result)
```
