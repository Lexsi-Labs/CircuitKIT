## 4  Config

Edit this cell. Everything below reads these names.


```python
# ── model ──────────────────────────────────────────────────────────────────
MODEL              = "Qwen/Qwen2.5-1.5B-Instruct"  # https://huggingface.co/Qwen/Qwen2.5-1.5B-Instruct
# MODEL            = "Qwen/Qwen2.5-0.5B-Instruct"
TASK               = "ioi"
# ioi | sva | greater_than | gender_bias | capital_country | hypernymy
# double_io | boolq | glue | mmlu | winogrande | winogrande_mc
# truthfulqa | ifeval | wmdp | gsm8k
ALGORITHM          = "eap-ig"
# eap | eap-ig | acdc | ibcircuit | cdt | eap-ig-activations
# eap-clean-corrupted | eap-exact | atp-gd | eap-gp | relp | peap | eap-ifr
LEVEL              = "node"       # node | neuron
SCOPE              = "both"       # heads | mlp | both
SPARSITY           = 0.3
OUT_DIR            = "./ck_out"
IG_STEPS           = 5
N_EXAMPLES         = 32
BATCH_SIZE         = 1
MLP_HOOK           = "mlp_out"    # mlp_out | post_act
CHAT_TEMPLATE_MODE = "auto"       # auto | on | off
DO_EVALUATE        = False        # --evaluate
DO_RANDOM          = False        # --random
```

## 5  Build


```python
import shlex
from pathlib import Path
Path(OUT_DIR).mkdir(parents=True, exist_ok=True)
cmd = (
    f"circuitkit discover -m {MODEL} -t {TASK} -a {ALGORITHM} "
    f"--level {LEVEL} --sparsity {SPARSITY} --scope {SCOPE} "
    f"--ig-steps {IG_STEPS} --batch-size {BATCH_SIZE} "
    f"--num-examples {N_EXAMPLES} --mlp-hook {MLP_HOOK} "
    f"--chat-template-mode {CHAT_TEMPLATE_MODE} "
    f"--output {OUT_DIR}/cli.pt"
)
if DO_EVALUATE:
    cmd += " --evaluate"
if DO_RANDOM:
    cmd += " --random"
print(cmd)
```

## 6  Run


```python
import shlex
from pathlib import Path
Path(OUT_DIR).mkdir(parents=True, exist_ok=True)
cmd = (
    f"circuitkit discover -m {MODEL} -t {TASK} -a {ALGORITHM} "
    f"--level {LEVEL} --sparsity {SPARSITY} --scope {SCOPE} "
    f"--ig-steps {IG_STEPS} --batch-size {BATCH_SIZE} "
    f"--num-examples {N_EXAMPLES} --mlp-hook {MLP_HOOK} "
    f"--chat-template-mode {CHAT_TEMPLATE_MODE} "
    f"--output {OUT_DIR}/cli.pt"
)
if DO_EVALUATE:
    cmd += " --evaluate"
if DO_RANDOM:
    cmd += " --random"
print(cmd)
import subprocess
subprocess.check_call(shlex.split(cmd))
```
